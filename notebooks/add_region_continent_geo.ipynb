# Combine region geometry to continent

```python
import geopandas as gpd

# Load your region-level GeoJSON
region_gdf = gpd.read_file("../data/processed/geojson/region_clean.geojson")

# --- Add Continent Mapping ---
# Manually map regions to continents
region_to_continent = {
    "North America": "Americas",
    "Central America": "Americas",
    "Caribbean and Bermuda": "Americas",
    "South America": "Americas",

    "Western Europe": "Europe",
    "Eastern Europe": "Europe",
    "Northern Europe": "Europe",
    "Southern Europe": "Europe",

    "Western Africa": "Africa",
    "Eastern Africa": "Africa",
    "Northern Africa": "Africa",
    "Central Africa": "Africa",
    "Southern Africa": "Africa",

    "Eastern Asia": "Asia",
    "Southeast Asia": "Asia",
    "Southern Asia": "Asia",
    "West Central Asia and the Middle East": "Asia",

    "Oceania": "Oceania",
    "Antarctica and Adjacent Islands": "Antarctica and Adjacent Islands",

    "Other": "Other"
}

# Apply mapping
region_gdf["Continent"] = region_gdf["ADMIN"].map(region_to_continent)

# Drop any unmatched (optional)
region_gdf = region_gdf.dropna(subset=["Continent"])

# --- Group and dissolve geometries by continent ---
continent_gdf = region_gdf.dissolve(by="Continent", as_index=False)

# Rename for consistency with your app
continent_gdf = continent_gdf.drop(columns=["ADMIN"])
continent_gdf.rename(columns={"Continent": "ADMIN"}, inplace=True)

continent_gdf
```

```python
# Save result
continent_gdf.to_file("../data/processed/geojson/continent_clean.geojson", driver="GeoJSON")

print("Continent GeoJSON saved as 'continent_clean.geojson'")
```
