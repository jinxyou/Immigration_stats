```python
def classify_birthplace(place):
    if place == "Total – Place of birth":
        return "Total"
    elif place == "Inside Canada":
        return "Inside Canada"
    elif place == "Outside Canada":
        return "Outside Canada"
    elif place in [
        "Newfoundland and Labrador", "Prince Edward Island", "Nova Scotia", "New Brunswick",
        "Quebec", "Ontario", "Manitoba", "Saskatchewan", "Alberta", "British Columbia",
        "Yukon", "Northwest Territories", "Nunavut"
    ]:
        return "Province"
    elif place in [
        "Americas", "Europe", "Africa", "Asia", "Oceania"
    ]:
        return "Continent"
    elif place in [
        "Northern Africa", "Western Africa", "Eastern Africa", "Middle Africa", "Southern Africa",
        "Eastern Asia", "Southeast Asia", "Southern Asia", "Western Asia", "Central Asia",
        "Western Europe", "Northern Europe", "Eastern Europe", "Southern Europe",
        "North America", "Central America", "Caribbean and Bermuda", "South America",
        "West Central Asia and the Middle East",
        "Central Africa", "Southern Africa", "Eastern Africa", "Northern Africa", "Western Africa",
        "Australia and New Zealand", "Melanesia", "Micronesia", "Polynesia"
    ]:
        return "Region"
    elif place in ["Other"]:
        return "Other"
    else:
        return "Country"
```

```python
pruid_to_province = {
    '10': 'Newfoundland and Labrador',
    '11': 'Prince Edward Island',
    '12': 'Nova Scotia',
    '13': 'New Brunswick',
    '24': 'Quebec',
    '35': 'Ontario',
    '46': 'Manitoba',
    '47': 'Saskatchewan',
    '48': 'Alberta',
    '59': 'British Columbia',
    '60': 'Yukon',
    '61': 'Northwest Territories',
    '62': 'Nunavut'
}

def map_province(dguid):
    csduid = dguid[9:]  # Extract the geographic unique identifier (CSDUID)
    pruid = csduid[:2]  # Extract the first 2 digits (PRUID)
    return pruid_to_province.get(pruid, 'Unknown')

```

```python
import pandas as pd

input_csv = "../data/raw/immigration_data/immigration_stats_census_subdivisions.csv"
output_parquet = "../data/processed/immigration_data/immigration_stats_census_subdivisions.parquet"
target_col = "Immigrant status and period of immigration (11):Total - Immigrant status and period of immigration[1]"

chunksize = 100_000
filtered_chunks = []

for chunk in pd.read_csv(input_csv, chunksize=chunksize):
    chunk[target_col] = pd.to_numeric(chunk[target_col], errors='coerce')
    filtered = chunk[(chunk[target_col] != 0) & (~chunk[target_col].isna())]
    filtered["Type"] = filtered["Place of birth (290)"].apply(classify_birthplace)
    filtered['Province'] = filtered['DGUID'].apply(map_province)
    filtered_chunks.append(filtered)

df = pd.concat(filtered_chunks, ignore_index=True)
df.to_parquet(output_parquet, index=False)
```
